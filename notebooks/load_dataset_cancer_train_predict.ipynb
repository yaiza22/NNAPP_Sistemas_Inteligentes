```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
from sklearn.model_selection import train_test_split

data = load_breast_cancer()
df_data= pd.DataFrame(data.data, columns=data.feature_names)
df_data.head()

df_target = pd.DataFrame(data.target, columns=['target'])
df_target.head()

df = pd.concat([df_data, df_target], axis=1)
df.head()

X_train, X_pred, y_train, y_pred = train_test_split(df_data, df_target, test_size=0.05, random_state=42)

df_train = pd.concat([X_train, y_train], axis=1)
df_train.to_csv("breast_cancer_train.csv", index=False)

# create dataframe from X_pred only
df_pred = pd.DataFrame(X_pred, columns=df_data.columns)
df_pred.to_csv("breast_cancer_pred.csv", index=False)

df_pred_result = pd.DataFrame(y_pred)
df_pred_result.to_csv("breast_cancer_results.csv", index=False)
```
